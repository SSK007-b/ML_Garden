```python
import yaml
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
from xgboost import XGBClassifier
```

```python
with open("D:\Github Sanity\ML Garden\projects\loan_approval\config.yaml", "r") as file:
    config_data = yaml.safe_load(file)
```

```python
config_data
```

```python
data = pd.read_csv(config_data["dataset_path"])
data.head()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data.columns = data.columns.str.strip().str.lower()
```

```python
def labelling(x):
    data[x] = LabelEncoder().fit_transform(data[x])
    return data
```

```python
for cat in config_data.get("categorical_features"):
    labelling(cat)
```

```python
X = data[config_data["independent_features"]]
y = data[config_data["dependent_features"]]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = config_data.get("test_size"), random_state = config_data.get("random_state"))
```

```python
model = XGBClassifier()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
print("Accuracy is ", accuracy_score(y_test, y_pred))
```
